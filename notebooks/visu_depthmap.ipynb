```python
import numpy as np
import matplotlib.pyplot as plt
import os
import glob
```

```python
path = "C:/Users/pokem/OneDrive/Documents/TRAVAIL/Code/PFE/outputs/pseudo_labels_v2/"
files = glob.glob(os.path.join(path, "*.npy"))
print(f"Found {len(files)} depth map files in {path}")
for file in files:
    depth_map = np.load(file)
    plt.imshow(depth_map, cmap='plasma')
    plt.colorbar()
    plt.title(f"Depth Map from {os.path.basename(file)}")
    plt.show()
```
